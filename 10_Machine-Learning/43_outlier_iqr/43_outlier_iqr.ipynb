## ******IQR (Interquartile Range) Based Outlier Detection******

Jab aapka data **Normally Distributed nahi hota**, yaani data **Skewed (Right-skewed ya Left-skewed)** hota hai, tab Z-score method fail ho jata hai. Waha hum **IQR Method** use karte hain.

---

## ******⚠️ Golden Rule (Kab Use Karna Hai?)******

* **Skewed Distribution:** Agar histogram me tail ek taraf kheench rahi hai ya extreme values mean ko shift kar rahi hain, tab IQR method best choice hai.
* Yeh method **Median** aur **Quartiles** par based hota hai, isliye extreme outliers se bilkul affect nahi hota (Robust to outliers).

---

## ******📊 1. Percentile Kya Hota Hai?******

Percentile ka simple matlab yeh hai ki **"Kitne percent log ya data points aapke barabar ya aapse peeche hain."**

> **Example:** Agar aapne bola ki **100th percentile par value 89 hai**:
> * Iska matlab dataset ka **100% data 89 ya 89 se chota hai**.
> * 89 aapke dataset ki **Maximum value** hai.
> * Agar kisi test me aapka **75th percentile** aaya, toh iska matlab 75% bachhon ke marks aapse kam ya barabar hain, aur bas 25% bachhe aapse aage hain.

---

## ******📦 2. Box Plot & Five-Number Summary******

Box Plot data ko visual summarize karne ka graphical tool hai, jo purely **Five-Number Summary** ke upar based hota hai:

| Summary Point | Name | Percentile | Meaning |
| :--- | :--- | :---: | :--- |
| **Minimum** | Lower Limit | $-$ | $Q_1 - 1.5 \times \text{IQR}$ (Isse choti value outlier ban jaati hai) |
| **$Q_1$** | First Quartile | **25th** | Pure data ka 25% hissa iske neeche rehta hai |
| **Median ($Q_2$)** | Second Quartile | **50th** | Pure data ka exact center (50% data chota, 50% bada) |
| **$Q_3$** | Third Quartile | **75th** | Pure data ka 75% hissa iske neeche rehta hai |
| **Maximum** | Upper Limit | $-$ | $Q_3 + 1.5 \times \text{IQR}$ (Isse badi value outlier ban jaati hai) |

---

## ******📏 3. What is IQR (Interquartile Range)?******

IQR basically aapke middle 50% data ka spread (width) hota hai:

$$\text{IQR} = Q_3 - Q_1$$

* Yeh $75^{\text{th}}$ percentile aur $25^{\text{th}}$ percentile ke beech ka gap hai.
* Jo center ka box banta hai, us box ki total length hi **IQR** hoti hai.

---

## ******🔍 Box Plot Ke Important Terms******

* **The Box (Interquartile Range):** 
  * Box ka bottom edge $Q_1$ hota hai aur top edge $Q_3$. Is rectangle box ke andar dataset ka **middle 50% data** band hota hai.
* **Median Line (Yellow/Center line):** 
  * Box ke beech me jo horizontal line hoti hai wo 50th percentile (Median) dikhati hai.
* **Whiskers (Moochein / T-bars):** 
  * Box ke dono ends se jo dandi bahar nikalti hai, unhe **Whiskers** kehte hain. 
  * Yeh data ke normal spread ko represent karti hain (Lower limit aur Upper limit tak jaati hain).
* **Outlier Points (Dots / Circles):** 
  * Jo points Whiskers ki boundary ($Q_1 - 1.5 \times \text{IQR}$ ya $Q_3 + 1.5 \times \text{IQR}$) ke bahar nikal jaate hain, unhe standalone dots se draw kiya jata hai. Wahi hamare actual outliers hote hain.

---

## ******🚨 IQR Outlier Detection Boundaries Formula******

$$\text{Lower Limit} = Q_1 - 1.5 \times \text{IQR}$$

$$\text{Upper Limit} = Q_3 + 1.5 \times \text{IQR}$$

$$\text{Rule: Agar koi value } X < \text{Lower Limit} \text{ ya } X > \text{Upper Limit} \implies \text{Outlier!}$$
