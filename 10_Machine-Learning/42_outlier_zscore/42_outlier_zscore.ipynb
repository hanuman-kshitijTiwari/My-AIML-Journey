## ******📌 Z-Score Outlier Removal Technique******

Outlier detection aur treatment ki sabse famous aur standard technique **Z-Score Method** hai. 

---

******⚠️ Core Assumption (Sabse Badi Shart)******

Yeh technique blindfolded hoke kisi bhi column par nahi lagayi ja sakti:
* **Assumption:** Aap jis bhi feature/column par Z-score method laga rahe ho, wo **Normally Distributed (Gaussian Distribution)** hona chahiye ya fir **Almost Normal Distribution** jaisa dikhna chahiye (Bell-shaped curve).
* Normal distribution me data ka nature yeh hota hai ki:
  * Maximum values **Centre (Mean)** ke aas-paas cluster hoti hain.
  * Jaise-jaise aap extreme Left (bohot kam) ya extreme Right (bohot zyada) jaate ho, data points ki sankhya drastically kam hoti chali jaati hai.

---

******📐 Empirical Rule (68-95-99.7 Rule)******

Normal distribution ke andar standard deviation ($\sigma$) aur mean ($\mu$) ke hisab se data mathematically predefined brackets me bat-ta hai:

1. **Mean $\pm$ $1\sigma$ Range:**
   $$\mu - 1\sigma \quad \text{se} \quad \mu + 1\sigma$$
   * Aapke pure data ki lagbhag **68.2%** values isi range ke andar aati hain.

2. **Mean $\pm$ $2\sigma$ Range:**
   $$\mu - 2\sigma \quad \text{se} \quad \mu + 2\sigma$$
   * Agar range ko thoda aur expand karein, toh lagbhag **95.4%** data points is bracket me cover ho jaate hain.

3. **Mean $\pm$ $3\sigma$ Range:**
   $$\mu - 3\sigma \quad \text{se} \quad \mu + 3\sigma$$
   * Is range me pure data ka lagbhag **99.7%** hissa cover ho jata hai.

---

******🚨 Outlier Rule (Close-to-Normal Checking)******

Statistical consensus ke hisab se:
* Jab **99.7%** population already $\mu \pm 3\sigma$ ke andar exist karti hai, toh jo bacha hua **0.3%** data extreme boundary ke bahar (chahe extreme left me ho ya extreme right me) chala gaya, use hum normal variations nahi maante.
* Unhe directly **Outliers** treat kiya jata hai.

$$\text{Lower Limit} = \mu - 3\sigma$$

$$\text{Upper Limit} = \mu + 3\sigma$$

$$\text{Condition: Agar } X < \text{Lower Limit} \text{ ya } X > \text{Upper Limit} \implies \text{Outlier!}$$

---

## ******❓ Is Technique Ko "Z-Score" Kyu Bolte Hain?******

Aapke dimag me sawal aayega ki jab hum limits mean aur standard deviation se nikaal rahe hain, toh iska naam Z-Score kyu pada?

Kyunki mathematically, kisi bhi data point ka standard position find karne ka formula hi **Z-Score** hota hai:

$$Z = \frac{X - \mu}{\sigma}$$

* **$X$:** Current data point
* **$\mu$:** Column ka Mean
* **$\sigma$:** Column ka Standard Deviation

#### 💡 Real Meaning:
* Z-score calculation ka simple matlab hota hai: **"Yeh particular value mean ($\mu$) se kitne Standard Deviations ($\sigma$) doori par khadi hai?"**
* Jab hum bolte hain ki value $\mu + 3\sigma$ se badi hai, toh uska Z-score calculate karne par wo seedha $Z > 3$ aayega.
* Jab value $\mu - 3\sigma$ se choti hai, toh uska Z-score $Z < -3$ aayega.

Isliye hum har value ka Z-score nikaalte hain aur direct check laga dete hain:

$$|Z| > 3 \implies \text{Outlier}$$

Yeh check karna aur normal distribution ki limits set karna dono mathematically 100% same cheez hain!

## ******Z-Score Based Outlier Treatment: Trimming vs Capping******

Jab Z-Score ($Z = \frac{X - \mu}{\sigma}$) calculate karke hume pata chal jata hai ki kaun-kaun se points boundaries ($\mu \pm 3\sigma$) ke bahar hain, tab unhe treat karne ke do standard approaches aate hain.

---

****** 🧪 Toy Dataset Example ******

Maan lo hamare paas ek test score ka normally distributed dataset hai:

$$\text{Data} = [48, 50, 52, 55, 60, 62, 65, 110, 10]$$

* **Mean ($\mu$):** $56.88$ (lagbhag $57$)
* **Standard Deviation ($\sigma$):** lagbhag $12$

#### ***Boundaries Calculate Karte Hain:***
* **Upper Limit ($\mu + 3\sigma$):** $57 + (3 \times 12) = 57 + 36 = \mathbf{93}$
* **Lower Limit ($\mu - 3\sigma$):** $57 - (3 \times 12) = 57 - 36 = \mathbf{21}$

> **Check:** 
> * $110 > 93$ $\longrightarrow$ **Outlier** (Extreme high)
> * $10 < 21$ $\longrightarrow$ **Outlier** (Extreme low)
> * Baaki sabhi points ($48, 50, 52, 55, 60, 62, 65$) range ke andar hain.

---

#### ***Dataset After Trimming:***
* $10$ drop ho gaya (kyunki $10 < 21$).
* $110$ drop ho gaya (kyunki $110 > 93$).

$$\text{Treated Data} = [48, 50, 52, 55, 60, 62, 65]$$

* **Fayda:** Dataset ka original shape (total row count) barabar rehta hai, koi information drop nahi hoti.
* **Nuksan:** Extreme points artificial limits par bundle ho jate hain, jisse tails par thodi artificial concentration ban sakti hai.

| **Action on Extreme Values** | Outlier points delete hote hain | Outlier points boundary values se replace hote hain |
| **Best Case** | Outlier clear-cut error/garbage entry ho | Dataset chhota ho aur data loss afford na ho sake |
