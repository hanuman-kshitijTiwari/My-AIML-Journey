****Outlier detection ka teesra sabse simple aur powerful approach **Percentile Method** hai. Isme hum distribution ke shape (normal ya skewed) par depend nahi karte, balki directly ranking aur percentages ke through cutoffs set karte hain.****

---

## ******🧠 Percentile Ka Funda (Core Concept)******

Percentile ka simple matlab hota hai: **"Data me kitne percent log ya values aapke peeche (equal ya less than) khade hain."**

* **1. 99th Percentile:**
  * Iska matlab dataset ka **99% data aapke barabar ya aapse chota hai**, aur sirf top **1%** log aapse aage hain.
* **2. 50th Percentile (Median):**
  * Iska matlab **50% log aapse peeche hain aur 50% log aapse aage hain**. Yeh data ka exact midpoint hota hai, isliye **50th Percentile = Median** kehlata hai.
* **3. Maximum Value ($95$):**
  * Agar dataset me max score $95$ hai, toh yeh **100th Percentile** kehlayega kyunki $100\%$ data points iske barabar ya isse kam hain.
* **4. Minimum Value ($10$):**
  * Agar dataset me lowest score $10$ hai, toh yeh **0th Percentile** kehlayega kyunki $0\%$ log iske peeche hain (ye sabse shuruat ka point hai).

---

## ******⚙️ Step-by-Step Outlier Detection Process******

Maan lo aapke paas kisi feature (jaise **Height**, **Weight**, ya **Salary**) ka data hai aur aapko usme outliers find karne hain:

1. **Threshold Amount Decide Karo:**
   * Aap distribution ke dono tails (extreme ends) se ek fix percentage buffer choose karte ho.
   * *Example:* Maan lo humne dono side se **1%** margin fix kiya.

2. **Boundaries Set Karo:**
   * **Upper Limit:** $99^{\text{th}}$ percentile se upar ki saari values ko **Outliers** maana jayega.
   * **Lower Limit:** $1^{\text{st}}$ percentile se neeche ki saari values ko **Outliers** maana jayega.

$$\text{Lower Limit} = \text{Percentile}(1\%)$$

$$\text{Upper Limit} = \text{Percentile}(99\%)$$

$$\text{Outlier Check: } X < \text{Lower Limit} \quad \text{ya} \quad X > \text{Upper Limit}$$

---

## ******🛠️ Outlier Treatment (Dono Approaches)******

Jab boundaries calculate ho jaati hain, tab humare paas wahi do main options hote hain:

* **✂️ 1. Trimming (Direct Removal):**
  * Jo records $1^{\text{st}}$ percentile se chote hain ya $99^{\text{th}}$ percentile se bade hain, un pure rows ko dataset se permanently delete kar do.
  ```python
  # Trimming Code
  lower_limit = df['height'].quantile(0.01)
  upper_limit = df['height'].quantile(0.99)
  df_trimmed = df[(df['height'] >= lower_limit) & (df['height'] <= upper_limit)]

____



## *****2. Capping (Winsorization):*****
* Data delete nahi karte. Jo values $1^{\text{st}}$ percentile se choti hain unhe $1^{\text{st}}$ percentile ki value bana dete hain, aur jo $99^{\text{th}}$ percentile se badi hain unhe $99^{\text{th}}$ percentile ki value par cap kar dete hain.


```python
import numpy as np

# Capping Code
lower_limit = df['height'].quantile(0.01)
upper_limit = df['height'].quantile(0.99)
df_capped = df.copy()
df_capped['height'] = np.clip(df_capped['height'], lower_limit, upper_limit)

```
